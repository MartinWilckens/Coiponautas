# Pancho

## ¿Qué es un análogo planetario?

**Definición.** Lugar o instalación en la Tierra con similitudes físicas (naturales o construidas) con un ambiente de la Luna, Marte o el espacio. Se usa para estudiar procesos planetarios y probar instrumentos, tecnologías y operaciones antes de volar[1][2].

**Tipos**
- *Naturales:* desiertos, regiones polares, terrenos volcánicos, cráteres de impacto, cuevas y tubos de lava. [1]
- *Construidos:* cámaras de altitud, simuladores de gravedad reducida (ARGOS), hábitats como CHAPEA (estadías de un año simulando Marte). [2]

**Para qué sirven**
- *Ciencia:* entender procesos geológicos, climáticos o biológicos parecidos a los de la Luna o Marte. [1]
- *Ingeniería y operaciones:* probar equipos, métodos y procedimientos de misión. [1]
- *Factores humanos:* la NASA estudia cinco riesgos del vuelo humano: radiación, aislamiento y confinamiento, distancia a la Tierra, gravedad y ambientes hostiles o cerrados. [2]

**Límites.** La Tierra no reproduce la gravedad reducida, el vacío lunar, la atmósfera delgada de Marte ni su ambiente de radiación. [1]

## ¿Qué tipos de análogo existen?

Foucher et al. clasifican los sitios análogos según su uso [3]:

| Tipo | Qué imita | Ejemplo | En nuestra lista |
|---|---|---|---|
| Analogía planetaria | Paisaje y morfología | Cráteres de impacto, campos volcánicos | Geológico |
| Procesos mecánicos y químicos | Erosión, polvo, hielo | Dunas, glaciares, alteración por agua | Ambiental o climático |
| Petrológico y mineralógico | Rocas y minerales | Basaltos, salares con sulfatos | Geológico |
| Interés astrobiológico | Habitabilidad, extremófilos, biofirmas | Desiertos hiperáridos, fuentes termales | Astrobiológico |
| Ingeniería | Condiciones para probar equipos y operaciones | Terreno para ensayar rovers o perforación | Operacional |

La NASA suma los **análogos de factores humanos**: aislamiento, confinamiento, distancia a la Tierra, gravedad y ambientes hostiles (p. ej. estaciones antárticas, CHAPEA) [2].

**Idea clave:** *análogo funcional*. Un sitio no es análogo en general; se elige por la propiedad que importa para un uso concreto [3].

## ¿Qué diferencia hay entre un análogo de campo y uno de laboratorio?

| | Campo | Laboratorio |
|---|---|---|
| **Qué es** | Sitio natural: desierto, cráter, glaciar | Instalación con condiciones controladas |
| **Ejemplos** | Atacama, Svalbard (AMASE), Arizona (NASA D-RATS) [3] | Cámaras de simulación, torres de caída, vuelos parabólicos, simulantes de regolito, hábitats cerrados (Mars-500, CHAPEA) [2][3] |
| **Qué reproduce** | Paisajes, escala y procesos naturales (erosión, polvo, hielo) que no se pueden recrear en laboratorio [3] | Temperatura, presión, humedad, polvo y radiación en cámara; gravedad reducida por segundos en torres de caída o vuelos parabólicos [3] |
| **Ventaja** | Realismo: prueba equipos, protocolos y al equipo humano en condiciones exigentes [3] | Control y repetición de variables; reproduce condiciones que el campo no tiene (vacío, presión de Marte) |
| **Límite** | No controlas las variables; sigue habiendo gravedad, atmósfera y protección contra la radiación de la Tierra | Escala pequeña; no reproduce el paisaje ni procesos complejos |

### ¿Cuándo usar cada uno?

- **Laboratorio:** cuando la condición clave no existe en ningún sitio terrestre (vacío, presión marciana, radiación, gravedad) o hay que aislar y repetir una variable. También en estudios psicológicos o de convivencia, donde la geología no importa (p. ej. Mars-500) [3].
- **Campo:** cuando importan el terreno, la escala o los procesos naturales, o se quiere probar todo junto: rover, instrumentos, comunicaciones y equipo humano [3].
- **Ambos combinados (lo más común):** las muestras se toman en terreno y se analizan en laboratorio; los equipos se prueban primero en cámara y después se validan en campo [3].

**Para el proyecto:** el reto pide sitios de campo. El laboratorio sirve para explicar qué condiciones necesitarían una cámara o un simulador porque la Tierra no las da, que es la parte de "dónde se rompe la analogía".
